```python
sales = [1000, 2500, 1500, 3000, 4500]

#Calculate total sales
total = sum(sales)
print(total) 
```

```python
#Calculate average sales
average = total / len(sales)

#Find highest sale
highest = max(sales)

#Find lowest sale
lowest = min(sales)

#Display sales greater than 2000
greater_sales = [x for x in sales if x > 2000]

# Display the results
print("Total Sales:", total)
print("Average Sales:", average)
print("Highest Sale:", highest)
print("Lowest Sale:", lowest)
print("Sales greater than 2000:", greater_sales)
```
