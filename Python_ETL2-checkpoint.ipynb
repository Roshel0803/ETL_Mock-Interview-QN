```python
import csv

# Read the CSV file
with open("D:\Talend_ETL_Project\employeesa5.csv", "r") as file:
    reader = csv.DictReader(file)

    total_salary = 0

    print("All Employee Records:")

    for row in reader:
        print(row)

        # Calculate total salary
        total_salary += int(row["salary"])

        # Display employees earning more than 50000
        if int(row["salary"]) > 50000:
            print("High Salary Employee:", row["name"],
                  "Salary:", row["salary"])

    # Display total salary
    print("\nTotal Salary:", total_salary)
```
